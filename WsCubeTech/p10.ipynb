### Function Transformation

converts non-normal distribution patter data to normal distribution pattern data 

```python
import pandas as pd
import seaborn  as sns
import matplotlib.pyplot as plt
```

```python
# working on origimal data
dataset = pd.read_excel(r"C:\Users\junai\Desktop\coding\ML\yahoo_data.xlsx")
dataset.head(5)
```

```python
dataset.isnull().sum()
```

```python
sns.distplot(dataset["Adj Close**"])
plt.show()
```

```python
# it has a long tail and may contain outlier , therefore its not in normal distribution
# u can first remove outlier and then convert it to normal distribution
# Using IQR method to remove outlier
q1 = dataset["Adj Close**"].quantile(0.25)
q3 = dataset["Adj Close**"].quantile(0.75)
IQR = q3-q1
min_r = q1 - ( 1.5 * IQR)
max_r = q3 + ( 1.5 * IQR)
min_r , max_r
```

```python
# since both + , roming outlier from both
dataset = dataset[dataset["Adj Close**"]<=max_r]
dataset = dataset[dataset["Adj Close**"]>min_r]
```

```python
sns.distplot(dataset["Adj Close**"])
plt.show()
```

```python
# no outliers visible now 
# now converting to normal data using sklearn
from sklearn.preprocessing import FunctionTransformer
import numpy as np
```

```python
ft = FunctionTransformer(func = np.log1p)
# or can use x^2
# ft = FunctionTransformer(func = lambda x :x**2)
```

```python
ft.fit(dataset[["Adj Close**"]])
```

```python
dataset["Adj Close**_tf_data"] =ft.transform(dataset[["Adj Close**"]])
dataset
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title("Before")
sns.distplot(dataset["Adj Close**"])

plt.subplot(1,2,2)
plt.title("After")
sns.distplot(dataset["Adj Close**_tf_data"])
plt.show()
```
