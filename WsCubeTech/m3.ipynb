## Multiple Linear Regression

We have to predict one isngle output based on multiple input

first we define the relationship between the input

if the relation between input is linear then we use multiple linear regression

if not then we perform feature selection 

equation)  y = m1x1 + m2x2 + ..+ mnxn + c


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
dataset = pd.read_csv(r"C:\Users\junai\Desktop\coding\ML\student_data.csv")
dataset.head()
```

```python
dataset.shape
```

```python
dataset.isnull().sum()
```

```python
# now will create graph of age vs salary and Experience vs salary to verify both follows linearity
sns.pairplot(data=dataset)
plt.show()
```

```python
sns.heatmap(data=dataset.corr(),annot=True)   # cheking correlation to see if the data is linear
plt.show()
```

```python
# seperating depenedent and independent variables
x = dataset.iloc[:,:-1]
y = dataset["Salary"]
```

```python
x.ndim   #already 2 dimensions
```

```python
# splitting train and test data
x_train , x_test , y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
# creating model
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
lr.score(x_test,y_test)   # testing accuracy  => 95%
```
