## Data Cleaning
### Finding null values

#### import libraries

```python
import pandas as pd
import seaborn as sns  # for graphs
import matplotlib.pyplot as plt
```

#### read data from file and read 3 records from datset

```python
dataset = pd.read_excel("yahoo_data.xlsx")
dataset.head(3)
```

#### count the no of data rows and cols

```python
dataset.shape
```

#### find null values

```python
dataset.isnull()
# dataset.notnull() -> give true for not  null values 
```

#### count total null values  in each col

```python
dataset.isnull().sum()
```

#### count % null values in each col

```python
(dataset.isnull().sum() / dataset.shape[0])*100  # shape[0] for total rows
```

#### count total null values in whole data

```python
dataset.isnull().sum().sum()
```

#### count total null values in data in each cell

```python
(dataset.isnull().sum().sum()/(dataset.shape[0]*dataset.shape[1]))*100
```

#### showing null values graphically 

```python
sns.heatmap(dataset.isnull())
plt.show()
```
