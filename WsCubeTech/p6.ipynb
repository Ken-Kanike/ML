## Label Encoding

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
#demo catgorial data
df = pd.DataFrame({"Fruits":['Apple', 'Banana', 'Cherry','Dragon Fruit','emon']})
df
```

```python
le = LabelEncoder()
le.fit_transform(df['Fruits'])
```

```python
df['en_fruits'] = le.fit_transform(df['Fruits'])
df
```
