```python
import pandas as pd
import seaborn as sns  # for graphs
import matplotlib.pyplot as plt
```

```python
# load dataset
dataset = pd.read_excel("C:/Users/junai/Desktop/coding/ML/uncleaned_data.xlsx")
dataset.head(10)
```

```python
dataset.shape
```

```python
dataset.isnull().sum()
```

```python
sns.heatmap(dataset.isnull())
plt.show()
```

#### dropping Open column since too much null values 

```python
dataset.drop(columns=['Open']) # deletes column and returns new dataset , does not make change to original data
```

```python
dataset.drop(columns=['Open'], inplace= True)  #permenantly deletes col from dataset
sns.heatmap(dataset.isnull())
plt.show()
```

#### deleting perticular rows with null values

```python
dataset.dropna()  # deletes all rows with null values and returns new dataset , does not make change to original data
```

```python
dataset.dropna(inplace=True) # deletes all rows with null values in original dataset
sns.heatmap(dataset.isnull())
plt.show()
```
