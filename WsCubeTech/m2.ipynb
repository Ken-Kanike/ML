## Simple Linear Progression

When there in only ONE INPUT FEATURE ( indepented variable ) , then we use simple linear regression

we use y = mx + c  equation to make predictions

m = (x2-x1)/(y2-y1)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
dataset = pd.read_csv(r"C:\Users\junai\Desktop\coding\ML\student_placements.csv")
dataset.head(5)
```

```python
#check for nuls
dataset.isnull().sum()

#need to check if data follows linearity
sns.scatterplot(x="CGPA",y="Package(LPA)",data=dataset)
plt.show()
```

We can see graph moving only in one direction that is upword , therefore graph follows linearity

```python
#input = CGPA  , Output = Package
# y = dependent , x = independent 
# finding which is x and which is y
# y = CGPA , x = Package
x = dataset[["CGPA"]]   # needs to be multi dimension
y = dataset["Package(LPA)"]  # single dimension needed
```

```python
#splitting data into train and test data
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size=0.2, random_state=42)  
# test size = 20%
# keep changing random state between 30 to 70 to get best accuracy
```

```python
#implementing mode
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
lr.fit(x_train,y_train) # modal training
```

```python
lr.score(x_test,y_test) #testing accuracy  => 89%
```

```python
lr.predict([[7.00]])  #testing 
```

```python
# getting some more values
lr.coef_   # m
```

```python
lr.intercept_  # c
```

```python
# how predict works
# using y = mx + c
2.79574049  * 7.00 + (-13.860675828519755)
```

```python
y_prd = lr.predict(x)
```

```python
#now seeing the actual prediction line
sns.scatterplot(x="CGPA",y="Package(LPA)",data=dataset)
plt.plot(dataset["CGPA"],y_prd,c="red")
plt.legend(["org data","predict line"])
plt.show()
```
