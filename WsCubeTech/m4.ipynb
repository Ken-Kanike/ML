##  Polynomial Regression

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv(r"student_data.csv")
dataset.head()
```

```python
plt.scatter(dataset["Experience"],dataset["Salary"])
plt.xlabel("Experrience")
plt.ylabel("Salary")
plt.show()
```

```python
# yaha to linear graph hai but iske lie non linear chahie but , mere pass data nhi tha to yahi use kar raha
# polynomial me generally curved paths bante hai data ke , archs ke form me 
#                      ..........     
#                .....           ....
#            ....                    ...
#        ...                            ....
#     ...
#    ...
#  ...
#...

#kuch aise 

# lets test how much data correlated to linearity
dataset.corr()
```

```python
# we can see its highly correlated linear data
# chalo convert karte hai ise polynomial feature me
# input and output alag karlo , depenedent and independent

x = dataset[["Experience"]]
y = dataset["Salary"]
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
pf =PolynomialFeatures(degree=2)      #^2  binomial  since we using 2 cols
pf.fit(x)
```

```python
x = pf.transform(x)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test , y_train , y_test =train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression   # uses linear model class only for polynomial
```

```python
lr = LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
lr.score(x_test,y_test)
```

```python
prd = lr.predict(x)
```

```python
plt.scatter(dataset["Experience"],dataset["Salary"])
plt.plot(dataset["Experience"],prd , c="red")
plt.xlabel("Experrience")
plt.ylabel("Salary")
plt.legend(["org","prd"])
plt.show()
```
