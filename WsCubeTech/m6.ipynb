# Classification

Classification is a **supervised learning task** where the goal is to **predict the class label** of new instances based on past observations.

- Used when the **target/output** variable is **discrete (categorical)**.
- Each record belongs to a class/group.
- The model learns from **labeled training data** to classify new examples.

### 📌 Real-Life Examples:
- Classifying emails as **spam or not spam**
- Classifying animals into **mammals, reptiles, birds**, etc.
- Identifying whether a **transaction is fraudulent**
- Predicting whether a **customer will churn**

---


### A) Classification Analysis
When the **target variable is categorical** (discrete in nature), and we want to assign data into predefined labels or groups, we perform **classification analysis**.

Classification can be broadly categorized into:

---

#### i) Binary Class Classification
- When the output variable has **two possible outcomes** such as:
  - Yes / No
  - Spam / Not Spam
  - Positive / Negative
- Common in:
  - Medical diagnosis (disease or no disease)
  - Loan approvals (approved / rejected)

#### ii) Multi-Class Classification
- The output variable has **more than two categories**:
  - Example: Classifying fruits as Apple, Mango, Banana
  - Handwritten digit recognition (0 to 9)

---

### 1) Linear Classification Algorithms

These assume **linear decision boundaries**, i.e., they separate classes using a straight line (in 2D), or a hyperplane (in higher dimensions).

#### i) Logistic Regression
- Despite the name, it’s used for **classification**, not regression.
- Uses the **sigmoid function** to map input values to probabilities.
- Ideal for binary classification.
- Output: probability between 0 and 1.

#### ii) Support Vector Machine (Linear Kernel)
- Finds the **optimal separating hyperplane** that maximizes margin between classes.
- Works well with linearly separable data.
- Can be extended to non-linear problems with **kernel trick**.

---

### 2) Non-Linear Classification Algorithms

These models **do not assume linear decision boundaries** and can handle complex data patterns.

#### i) k-Nearest Neighbour (KNN)
- A **lazy learning algorithm**.
- Classifies a data point based on the **majority vote** of its k nearest neighbors.
- Distance metric (like Euclidean) is used to find neighbors.

#### ii) SVM (with Non-linear Kernel)
- When data is not linearly separable, **non-linear kernels** (like RBF or polynomial) are used.
- Transforms data into higher dimensions to find a hyperplane.

#### iii) Naive Bayes
- Based on **Bayes’ Theorem** and assumes **feature independence**.
- Works well with text classification problems (e.g., spam filtering).
- Fast and efficient, even with large datasets.

#### iv) Decision Tree Classification
- A tree-like structure where nodes represent features, branches represent decisions, and leaves represent outcomes.
- Handles both numerical and categorical data.
- Prone to overfitting if not pruned.

#### v) Random Forest Classification
- An **ensemble method** built using multiple decision trees.
- Reduces overfitting by averaging predictions.
- Robust and performs well in many real-world applications.

> Note: You've written "Decision tree Regression" and "Random Forest Regression" — if used for classification, these should be *Decision Tree Classifier* and *Random Forest Classifier*.

### 🔍 Evaluating Classification Models

#### i) Log Loss (Cross-Entropy Loss)
- Measures the performance of a classification model whose output is a **probability value between 0 and 1**.
- Lower the value, better the model.
- Formula:
  \[
  \text{Log Loss} = -\frac{1}{N} \sum (y \cdot \log(\hat{y}) + (1 - y) \cdot \log(1 - \hat{y}))
  \]

#### ii) Confusion Matrix
- A 2x2 table used for evaluating binary classification:
  - **TP**: True Positives
  - **FP**: False Positives
  - **FN**: False Negatives
  - **TN**: True Negatives

- Useful metrics derived from it:
  - Accuracy = (TP + TN) / Total
  - Precision = TP / (TP + FP)
  - Recall = TP / (TP + FN)
  - F1 Score = 2 * (Precision * Recall) / (Precision + Recall)

#### iii) AUC - ROC Curve
- **ROC** = Receiver Operating Characteristic
- Plots **True Positive Rate vs False Positive Rate** at various thresholds.
- **AUC** = Area Under Curve; higher is better.
  - AUC = 1 → perfect classifier
  - AUC = 0.5 → random guess
