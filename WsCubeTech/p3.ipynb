```python
# performing all operations on real data 
dataset = pd.read_excel("C:/Users/junai/Desktop/coding/ML/catg_data.xlsx")

# Fill missing values using backward fill (fills NaN with the next available value in the column)
# dataset.fillna(method='bfill', inplace=True)

# # Fill missing values using forward fill (fills NaN with the previous available value in the column)
# dataset.fillna(method='ffill', inplace=True)

# # # Forward fill across columns (fills NaN using previous value in the same row)
# # dataset.fillna(method='ffill', axis=1, inplace=True)

# # # Backward fill across columns (fills NaN using next value in the same row)
# # dataset.fillna(method='bfill', axis=1, inplace=True)
# # Forward fill across columns (row-wise fill for NaN values)
# dataset = dataset.apply(lambda row: row.ffill(), axis=1)

# # Backward fill across columns (row-wise fill for NaN values)
# dataset = dataset.apply(lambda row: row.bfill(), axis=1)


# # Fill missing values in categorical columns with the most frequent (mode) value
# for col in dataset.select_dtypes(include='object').columns:
#     dataset[col].fillna(dataset[col].mode()[0], inplace=True)


# # above methods are deprecated in pandas 3.0
# Fill missing values using backward fill (column-wise)
import pandas as pd

# Fill missing values using backward fill (column-wise)
dataset = dataset.bfill()

# Fill missing values using forward fill (column-wise)
dataset = dataset.ffill()

# Fill NaN values row-wise using forward fill
dataset = dataset.T.ffill().T  # Transpose to operate row-wise

# Fill NaN values row-wise using backward fill
dataset = dataset.T.bfill().T  # Transpose again to revert

# Fill categorical missing values with the most frequent (mode) value
for col in dataset.select_dtypes(include='object').columns:
    dataset[col] = dataset[col].fillna(dataset[col].mode().iloc[0])  # Use iloc[0] to avoid Series warnings

# Convert object dtypes properly to avoid future downcasting issues
dataset = dataset.infer_objects(copy=False)

# Optionally, set Pandas option to avoid silent downcasting in the future
pd.set_option('future.no_silent_downcasting', True)

# Save the updated dataset back to Excel
dataset.to_excel("C:/Users/junai/Desktop/coding/ML/catg_data.xlsx", index=False)
```
