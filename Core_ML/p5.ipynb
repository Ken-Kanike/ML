## One Hot Encoding

#### Gender -> Male / Female 0 - 1

```python
import pandas as pd
```

```python
# load dataset
dataset = pd.read_excel(r"gender_cat_data.xlsx")
dataset.head(10)
```

#### 2 ways to perform one hot encoding
##### 1. get dummies 
##### 2. one hot encoder class

#### Get dummies

```python
en_data = dataset[["Gender","Married"]]
en_data
```

```python
pd.get_dummies(en_data)
```

```python
pd.get_dummies(en_data).info()
# need to convert to numbers instead of boolean
```

## 2. one hot encoder class

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder()
ohe.fit_transform(en_data)
```

```python
# <1000x4 sparse matrix of type '<class 'numpy.float64'>'
# 	with 2000 stored elements in Compressed Sparse Row format>
# we need to convert this to array format
ar = ohe.fit_transform(en_data).toarray()
# we need to convert this to dataframe format
```

```python
pd.DataFrame(ar,columns=["Gender_Female", "Gender_Male","Married_No", "Married_Yes"])
```

```python
# now the data is too much we can draw drop some data for repeated columns
ohe = OneHotEncoder(drop='first')
ohe.fit_transform(en_data)
ar = ohe.fit_transform(en_data).toarray()
pd.DataFrame(ar,columns=["Gender_Male", "Married_Yes"]) # drops first col for every en_data
```
