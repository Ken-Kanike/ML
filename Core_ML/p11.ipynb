### Feature Selection

selecting which cols of  data must be used for ml model

1) Forward Elimination
2) Backward Elimination

#### Forward Elimination

```python
# Col 1       i)if highest accurate is col3
#                            ii) if highest  accuracy of col3,col1 is highest  and > col3
#             col3           col3     iii) Now select the highest giving combo  and > col3,col1
# Col 2       col1           col2
#                            col1      
#             col3
# Col 3       col2           col3
#                            col2
#             col3           col4 
# Col 4       col4
```

#### Backword Elimination


```python
# 4   ->    3   ->   2->    1 ->
# opposit of forward elimination
```

### Feature selection technique using code

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from mlxtend.feature_selection import SequentialFeatureSelector

# Load dataset
dataset = pd.read_excel(r"yahoo_data.xlsx")

# Separate features and target
# Assuming last column is target
x = dataset.iloc[:, :-1]
y = dataset.iloc[:, -1]  # or dataset["Adj Close**"]

# 1. Drop any non-numeric columns like dates
# Option A: Drop specific column by name if exists
if 'Date' in x.columns:
    x = x.drop(columns=['Date'])

# Option B: Keep only numeric columns automatically
x = x.select_dtypes(include=['number'])

print(f"Features used for model: {x.columns.tolist()}")

# Initialize regression model (LinearRegression for continuous target)
lr = LinearRegression()

# Initialize SequentialFeatureSelector
fs = SequentialFeatureSelector(
    lr,
    k_features=min(6, x.shape[1]),  # select up to 6 or less if less features available
    forward=True, # false for backword elimination
    floating=False,
    scoring='r2',  # regression metric
    cv=5,
    n_jobs=-1
)

# Fit the feature selector
fs.fit(x, y)

# Print selected features
print("Selected features:", fs.k_feature_names_)
print("Cross-validation R2 score:", fs.k_score_)
```
