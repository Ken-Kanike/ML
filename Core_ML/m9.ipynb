# 📊 Handling Imbalanced Datasets in Machine Learning

In real-world datasets, **imbalanced classes** are very common — for example, fraud detection, medical diagnosis, and rare event prediction. This imbalance can bias the model towards the majority class and degrade performance on minority class predictions.

---

## 1️⃣ Imbalanced Dataset

### 🔍 Definition:
An **imbalanced dataset** refers to a classification problem where the number of instances in each class is **not approximately equal**.

For example:

| Class       | Count |
|-------------|-------|
| Positive (1)| 50    |
| Negative (0)| 950   |

This is a **highly imbalanced** dataset (5% positive, 95% negative).

### ❗ Why It’s a Problem:
- Accuracy may appear high, but the model might just be predicting the majority class.
- Minority class is often more **important** (e.g., detecting fraud, disease).
- Evaluation metrics like **accuracy become misleading**.

### 📉 Example:
If a model always predicts “0” (majority), in a 95%-5% dataset:
- Accuracy = 95% ✅
- But **Recall for class 1 = 0** ❌ (missed all positive cases)

---

## 2️⃣ Bias in ML Models

### 🔍 What is Bias?
**Bias** is a systematic error introduced by incorrect assumptions in the model or data. It leads to unfair or inaccurate outcomes, especially for underrepresented groups or classes.

### 🎯 Sources of Bias:
- **Data collection bias**: Data isn't representative of the real world
- **Label bias**: Labels are incorrect or skewed
- **Algorithmic bias**: Model favors certain patterns due to training imbalance

---

## 🔍 Bias Detection Techniques

### ✅ Statistical Checks:
- Analyze distribution of predictions across groups
- Confusion matrix per class or subgroup
- Use metrics like **precision, recall, F1-score per class**

### ✅ Visualization:
- Histograms or bar plots showing class distribution
- ROC curve, precision-recall curve per class

---

## ✅ Bias Removal Techniques

### ⚖️ Approaches:
| Method                    | Description |
|---------------------------|-------------|
| **Data-level**            | Rebalancing data via sampling (undersampling, oversampling) |
| **Algorithm-level**       | Penalizing majority class, boosting minority class |
| **Post-processing**       | Adjusting prediction thresholds to favor balance |

---

## 3️⃣ Random Under-Sampling

### 🔍 What is it?
**Random under-sampling** reduces the number of samples from the **majority class** to match the minority class.

### ⚙️ How it works:
- Randomly remove samples from the majority class
- Balance class counts

### ✅ Pros:
- Simple and fast
- Can improve minority class learning

### ❌ Cons:
- **Loses information** (potentially important majority data discarded)
- Can underperform on test data

### 📌 Example:
Original:
- Class 0: 950
- Class 1: 50

After under-sampling:
- Class 0: 50 (randomly selected)
- Class 1: 50

---

## 4️⃣ Random Over-Sampling

### 🔍 What is it?
**Random over-sampling** duplicates existing samples from the **minority class** to balance it with the majority class.

### ⚙️ How it works:
- Randomly replicate data points in the minority class
- Retain all majority class data

### ✅ Pros:
- Preserves all original data
- Helps model learn rare patterns

### ❌ Cons:
- May lead to **overfitting** (same minority examples repeated)
- Increases training time

### 📌 Example:
Original:
- Class 0: 950
- Class 1: 50

After over-sampling:
- Class 0: 950
- Class 1: 950 (by duplicating)

---

## 🧠 Choosing Between Under- and Over-Sampling

| Criteria                        | Under-Sampling            | Over-Sampling             |
|---------------------------------|---------------------------|---------------------------|
| Dataset size (large)            | ✅ Recommended             | ❌ Slower, large memory   |
| Dataset size (small)            | ❌ Loses data              | ✅ Safe, keeps all data   |
| Risk of overfitting             | ✅ Less                    | ❌ Higher risk            |
| Model training speed            | ✅ Faster                  | ❌ Slower (more data)     |

---

## 🧪 Real-world Use Cases of Imbalanced Datasets

- ✅ Medical diagnosis (rare disease)
- ✅ Fraud detection
- ✅ Credit scoring
- ✅ Defect detection in manufacturing
- ✅ Spam email detection

````python
# practical implementation
import pandas as pd
````

````python
dataset = pd.read_csv(r"subscription_data.csv")
dataset.head()
````

````python
#checking data imbalance 
dataset["Purchased"].value_counts()
````

````python
# we can see there is data imbalance , as 0 has 287 records and 1 has 192 which will make model bias towards 0 prediction

# seperate data
x = dataset.iloc[: , :-1]
y = dataset["Purchased"]
````

````python
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x,y,random_state=42,test_size=0.2)
````

````python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
````

````python
lr.fit(x_train,y_train)
````

````python
lr.score(x_test,y_test)
````

````python
# we have got low accuracy 
# now we have to check if its bias or not

lr.predict([[19,19000]])
````

````python
lr.predict([[56,115235]])
````

````python
# we can its giving 0 even for 1 result , which confirms data imbalace and bias

# now we need to data balance through
# random over balanceing and underbalancing

# we need one module for it and its , pip install imblearn

#1) Random Under Sampling
````

````python
import pandas as pd
from imblearn.under_sampling import RandomUnderSampler
````

````python
dataset = pd.read_csv(r"subscription_data.csv")
dataset.head()
````

````python
# seperate data
x = dataset.iloc[: , :-1]
y = dataset["Purchased"]
````

````python
ru = RandomUnderSampler()
ru_x , ru_y = ru.fit_resample(x,y)
ru_y.value_counts()
````

````python
# now we can see that data is be balanced properly

from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(ru_x,ru_y,random_state=42,test_size=0.2)
````

````python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
lr.fit(x_train,y_train)
````

````python
lr.score(x_test,y_test)
````

````python
# we can see a slight increase in accuracy

# now let us test data

lr.predict([[19,19000]])
````

````python
lr.predict([[56,115235]])
````

````python
# we can see modal is not bias now

#2) Random Over sampling
````

````python
import pandas as pd
from imblearn.over_sampling import RandomOverSampler
````

````python
dataset = pd.read_csv(r"subscription_data.csv")
dataset.head()

# seperate data
x = dataset.iloc[: , :-1]
y = dataset["Purchased"]
````

````python
ro = RandomOverSampler()
ro_x , ro_y = ro.fit_resample(x,y)
ro_y.value_counts()
````

````python
# now we can see that data is be balanced properly

from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(ro_x,ro_y,random_state=42,test_size=0.2)
````

````python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
lr.fit(x_train,y_train)
````

````python
lr.score(x_test,y_test)
````

````python
lr.predict([[19,19000]])
````

````python
lr.predict([[56,115235]])
````

````python
# we can see modal is not bias now
````
