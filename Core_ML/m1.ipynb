## Supervised machine learning
A) Regression learning

B) Classification learning

### A)Regression Analysis
When the data or outcome is continous in nature

#### 1) Linear Regresssion Algorithms

    i) Simple Linear Regression

    ii) Multi-Linear Regression

    iii) Lasso Regression

    iv) Ridge Regression



#### 2) Non-linear Regression Algorithms

    i) Polynomial Regression

    ii) Decision tree Regression

    iii) Random Forest Regression

    iv) Support Vector Machine

    v) k-Nearest Neaighbour

### A)Classification Analysis
When the data or outcome is descrete  in nature

#### i) Binary Class Classification
       - can be divied into yes no type classes

#### ii) Multi Class Classification
      - cannot be divied into yes no type classes , have multiple classes


#### 1) Linear Classification Algorithms

    i) Logistic Regression 

    ii)  Support Vector Machine


#### 2) Non-linear Classification Algorithms

    i) k-Nearest Neaighbour

    ii) SVM ( Support Vector Machine )

    iii) Naive Bayes

    iv) Decision tree Regression

    v) Random Forest Regression
